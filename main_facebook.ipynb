```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from datetime import datetime
from utils.helpers import generar_nombre_archivo
from scraping.extractor_facebook import (
    iniciar_sesion,
    navegar_a_pagina,
    recolectar_publicaciones_pagina,
    recolectar_links_publicaciones,
    guardar_links_csv,
)


try:
    service = Service(ChromeDriverManager().install())
    driver = webdriver.Chrome(service=service)

    ok = iniciar_sesion(
        driver,
        user="jamoncayop@gmail.com",
        pwd="Adaptiv3@*",
        espera_recaptcha=180
    )

    if not ok:
        raise RuntimeError("No se pudo iniciar sesión en Facebook.")

    # 1) Ir a la página del IMSS
    navegar_a_pagina(driver, "https://www.facebook.com/IMSSmx")

    # 2) Recolectar publicaciones (ajusta max_posts si quieres más/menos)
    #publicaciones = recolectar_publicaciones_pagina(driver, max_posts=20, max_scrolls=25)
    links = recolectar_links_publicaciones(driver, max_links=3, max_scrolls=25)
    
    # 3) Crear el nombre dinámico para el archivo
    fecha_actual = datetime.now().strftime("%Y-%m-%d")
    base_filename = f"db/clasif_facebook_{fecha_actual}_jaime.csv"

    # Generar un nombre único para el archivo
    nombre_archivo = generar_nombre_archivo(base_filename)

    # 4) Guardar los links en el archivo CSV con el nombre generado
    guardar_links_csv(links, nombre_archivo)

except Exception as e:
    print("❌ Error en la ejecución:", e)
finally:
    try:
        driver.quit()
    except:
        pass
```
